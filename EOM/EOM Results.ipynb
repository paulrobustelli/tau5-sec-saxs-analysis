# EOM: WT versus AA
Analysis of completed ATSAS-online jobs WT120EOM and AA120EOM, downloaded 2026-09-25.

Inputs: WT frames 162–178 with buffer 220–236; AA frames 168–188 with buffer 24–70 (zero-based, inclusive). Both use the 120-residue constructs. Each run has a 10,000-conformer RANCH pool; GAJOE uses 100 cycles, 1,000 generations, 50 ensembles, and exactly 50 non-repeated conformers per selected ensemble. No additive constant was fitted.

The returned fit covers **0.012–0.50 Å⁻¹**, although submitted data extend to 0.60 Å⁻¹. These plots show the returned range, without extrapolation.

The server's selected distribution and its reported average summarize selection across cycles; the 50 structures listed in the final log describe the best ensemble. Keep these distinct. Histogram bin coordinates give slightly different averages from the server's unbinned header values.


```python
from pathlib import Path
import re, json
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown
ROOT = Path.cwd()
colors = {'WT':'#172869', 'AA':'#c65f12'}
results={}; curves={}; histograms={}
for sample in ['WT','AA']:
    d=ROOT/'data'/sample; selected=d/'selected'
    log=(selected/'logFile_001_1.log').read_text()
    rows=re.findall(r'\d+\)\s+(\S+)\s+([\d.]+)\s+([\d.]+)\s+~([\d.]+)',log)
    rg=np.array([float(r[1]) for r in rows]); dm=np.array([float(r[2]) for r in rows])
    assert len(rg)==50
    pool=np.loadtxt(d/'ranch_size.csv',usecols=(0,1))
    fit=np.loadtxt(selected/'profiles_001_1.fit',skiprows=1)
    hist=np.loadtxt(selected/'Rg_distr_001_1.txt',skiprows=5)
    assert np.allclose(hist[:,1:].sum(axis=0),1)
    residual=(fit[:,1]-fit[:,3])/fit[:,2]
    header=(selected/'Rg_distr_001_1.txt').read_text().splitlines()
    result=dict(best_ensemble_mean_Rg_A=float(rg.mean()),best_ensemble_Rg_SD_A=float(rg.std()),
        best_ensemble_RMS_Rg_A=float(np.sqrt(np.mean(rg**2))),best_mean_Dmax_A=float(dm.mean()),
        pool_mean_Rg_A=float(pool[:,0].mean()),pool_Rg_SD_A=float(pool[:,0].std()),
        server_selected_distribution_mean_Rg_A=float(header[0].split('=')[1]),
        server_chi2=float(re.search(r'Chi\^2\s*:\s*([\d.]+)',log)[1]),
        mean_squared_standardized_residual=float(np.mean(residual**2)),
        q_min=float(fit[0,0]),q_max=float(fit[-1,0]),fit_points=len(fit),
        residual_mean=float(residual.mean()),residual_lag1=float(np.corrcoef(residual[:-1],residual[1:])[0,1]))
    results[sample]=result; curves[sample]=(fit,residual); histograms[sample]=(hist,rg)
(ROOT/'summary.json').write_text(json.dumps(results,indent=2))
display(Markdown('| Sample | Best ensemble mean Rg ± conformational SD (Å) | RMS Rg (Å) | EOM distribution mean (Å) | Reported χ² |\n|---|---:|---:|---:|---:|\n'+'\n'.join(f"| {s} | {r['best_ensemble_mean_Rg_A']:.2f} ± {r['best_ensemble_Rg_SD_A']:.2f} | {r['best_ensemble_RMS_Rg_A']:.2f} | {r['server_selected_distribution_mean_Rg_A']:.2f} | {r['server_chi2']:.3f} |" for s,r in results.items())))
```

```python
fig,ax=plt.subplots(2,2,figsize=(11,8),constrained_layout=True)
for j,s in enumerate(['WT','AA']):
    h,rg=histograms[s]; width=np.median(np.diff(h[:,0]))
    ax[0,j].stairs(h[:,1]/width,np.r_[h[:,0]-width/2,h[-1,0]+width/2],color='0.55',label='Starting pool (10,000)')
    ax[0,j].stairs(h[:,2]/width,np.r_[h[:,0]-width/2,h[-1,0]+width/2],color=colors[s],fill=True,alpha=.35,label='EOM selected distribution')
    ax[0,j].axvline(results[s]['server_selected_distribution_mean_Rg_A'],color=colors[s],ls='--')
    ax[0,j].set(title=s,xlabel='Rg (Å)',ylabel='Probability density (Å⁻¹)');ax[0,j].legend(fontsize=8)
    ax[1,0].hist(rg,bins=np.arange(12,65,3),density=True,histtype='step',lw=2,color=colors[s],label=s)
    h=histograms[s][0];ax[1,1].plot(h[:,0],h[:,2]/np.median(np.diff(h[:,0])),color=colors[s],label=s)
ax[1,0].set(title='Best 50-conformer ensembles',xlabel='Rg (Å)',ylabel='Probability density (Å⁻¹)');ax[1,0].legend()
ax[1,1].set(title='EOM selected distributions',xlabel='Rg (Å)',ylabel='Probability density (Å⁻¹)');ax[1,1].legend()
fig.savefig('EOM_Rg_distributions.png',dpi=200);fig.savefig('EOM_Rg_distributions.pdf');fig.savefig('EOM_Rg_distributions.svg');plt.show()
fig=plt.figure(figsize=(11,7),layout='constrained'); gs=fig.add_gridspec(2,2,height_ratios=[3,1])
for j,s in enumerate(['WT','AA']):
    a=fig.add_subplot(gs[0,j]);b=fig.add_subplot(gs[1,j],sharex=a);f,r=curves[s]
    a.errorbar(f[:,0],f[:,1],yerr=f[:,2],fmt='.',ms=2,elinewidth=.5,color=colors[s],alpha=.6,label='Measured ± propagated σ')
    a.plot(f[:,0],f[:,3],color='black',lw=1.2,label='EOM best ensemble')
    a.set(yscale='log',ylabel='I(q) (input intensity units)',title=f"{s}: reported χ² = {results[s]['server_chi2']:.3f}");a.legend(fontsize=8)
    b.plot(f[:,0],r,'.',ms=2,color=colors[s]);b.axhline(0,color='black',lw=.7);b.axhspan(-2,2,color='0.8',alpha=.3)
    b.set(xlabel='q (Å⁻¹)',ylabel='(data−fit)/σ')
fig.savefig('EOM_fits.png',dpi=200);fig.savefig('EOM_fits.pdf');fig.savefig('EOM_fits.svg');plt.show()
```

## Interpretation and limits
The best ensembles have mean Rg approximately **26.7 Å (WT)** and **32.0 Å (AA)**. The server selected-distribution means are **27.40 Å** and **31.68 Å**. AA therefore favors larger conformations under this common EOM modeling protocol.

The ± values above describe **conformational spread**, not uncertainty in the mean, not a confidence interval, and not uncertainty from buffer subtraction. The best-ensemble RMS sizes, sqrt(mean(Rg²)), are also reported: they are the more relevant ensemble quantity for the low-q limit when conformers have equal I(0). Model-coordinate Rg and experimental scattering Rg need not match exactly because hydration and scattering contrast enter the latter.

These are single-chain ensembles by construction. Fits are compatible with a monomer ensemble, but do not independently determine molecular weight or rule out oligomer contamination. RANCH pools and selected populations are model-dependent; an apparent multimodal histogram is not proof of discrete physical states. One random pool per sample was used; the 100 optimizer cycles do not quantify independent-pool or experimental uncertainty.

The reported χ² is retained verbatim. The independently computed mean squared standardized residual uses N in the denominator and is listed separately in summary.json; it is not assigned a formal number of degrees of freedom. Error bars preserve the supplied errors. Shared buffer errors can correlate data; the diagonal residual analysis does not model full covariance. AA has structured residuals: the fit overpredicts much of q ≈ 0.23–0.36 Å⁻¹ and underpredicts the high-q tail. Its lag-one residual correlation is about 0.55 (WT: approximately 0.00). This flags systematic mismatch that may reflect the conformer pool, scattering model, or background; it does not identify oligomers. Inspect residual structure rather than using χ² alone.
