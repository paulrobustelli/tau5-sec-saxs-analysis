# Results comparison: exported I(q), Guinier and extended Guinier
This notebook reads the produced scattering curves; it does not rerun baseline selection or EFA. Every curve is shown **I(q) first**, followed by conventional/extended overlays and signed residuals. The curve registry retains source paths, sample, error assumptions, and hashes.

Start with the leading-edge section: it tests direct averages before substantial fitted overlap. The additional component is not automatically an oligomer or noise. EFA support constraints can force a contribution to zero; low fitted overlap is model-dependent.

Conventional fits use the widest tested range with qmax Rg ≤1.1, extended fits ≤2. The selector enforces approximate validity limits, not a best-fit guarantee; inspect residuals and qmin sensitivity. qmin=0.012 Å⁻¹ is primary. The matched-range extended column separates changing the functional form from extending the range. Original fixed 0.012–0.030 fits are retained in JSON and flagged if outside the IDP limit.

Extended model: ln(I/I0)=−(qRg)²/3+0.0479(ν−0.212)(qRg)⁴, with Rg coupled to ν by Zheng–Best equation 6 (120 residues,119 peptide bonds). This is a single-chain hypothesis, especially tentative for the WT shoulder. Conventional fitting uses weighted log intensity; extended fitting uses weighted intensity, so their chi-square values have different objectives. [Zheng & Best,2018](https://doi.org/10.1016/j.jmb.2018.03.007).

Uncertainties: the ± values are local conditional fit SEs; Monte Carlo intervals propagate exported errors at fixed q limits. EFA main/average use their existing paired perturbation curves. Direct-window comparisons propagate shared-background covariance, including overlapping frames. Baseline/support/q-range sensitivity is a separate systematic uncertainty, **not** pooled into statistical error bars. Unknown q covariance, uncertainty in data-selected edges, and unrecorded upstream processing are not reconstructed. No claim of a purified monomer follows from a good fit.
